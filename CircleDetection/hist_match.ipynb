```python
import cv2
import numpy as np
from skimage.exposure import match_histograms
import matplotlib.pyplot as plt

# Read the image 
img1 = cv2.imread('sharpieAndHigh1.jpg')
img2 = cv2.imread('sharpieAndHigh2.jpg')
```

```python
# Convert BGR to RGB if you're using OpenCV
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
```

```python
matched = match_histograms(img2_rgb, img1_rgb, channel_axis=-1)
```

```python
matched_bgr = cv2.cvtColor((matched).astype(np.uint8), cv2.COLOR_RGB2BGR)
```

```python
# Convert from BGR to RGB for display
result_rgb = cv2.cvtColor(matched_bgr, cv2.COLOR_BGR2RGB)

# Show images
plt.figure(figsize=(15,5))
plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Original Image 1")

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Original Image 2")

plt.subplot(1, 3, 3)
plt.imshow(result_rgb)
plt.title("Image 2 color matched to image 1")

plt.show()
```

```python
cv2.imwrite("sharpieAndHigh2Matched.jpg", matched_bgr)
```
